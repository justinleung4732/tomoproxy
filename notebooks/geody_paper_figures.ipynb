```python
# Functions to run calculations
def seismic_sshells_200(fileloc, comp, ppv_model_type, min_model = '', seismic_model = ''):
    """
    Function to create LayeredModels where thermochemical models only have chemically distinct
    material in the bottom 200km of the mantle.    
    
    Parameters
    -------
    fileloc: str
        A directory of where the LayeredModels of the thermal and thermochemical models are stored.
    comp: str
        The name of the thermochemical composition. Should be either "BMO", "MORB" or "HC".
    ppv_model_type: str
        The name of the pPv stability scenario. Should be either "noppv", "ppv" or "partppv".
    min_model: str
        The mineralogical model used. Should either be "SLB_2022" or "SLB_2011".
    seismic_model: str
        The name of the seismic model used in the filename.
    
    Returns
    -------
    Vp_layer: LayeredModel class object
        The spherical harmonic coefficients of a Vp model at specific layer depths.
    Vs_layer: LayeredModel class object
        The spherical harmonic coefficients of a Vs model at specific layer depths.
    Vc_layer: LayeredModel class object
        The spherical harmonic coefficients of a Vc model at specific layer depths.
    """
    filename = os.path.join(fileloc, f'SH_{seismic_model}_{comp}_{ppv_model_type}_{min_model[-2:]}')

    Vp_layer = lm.LayeredModel(f'{filename}_Vp')
    Vs_layer = lm.LayeredModel(f'{filename}_Vs')
    Vphi_layer = lm.LayeredModel(f'{filename}_Vc')
    
    if comp != 'pyrolite':
        Vp_TH = lm.LayeredModel(f'{filename}_Vp'.replace(comp, 'pyrolite'))
        Vs_TH = lm.LayeredModel(f'{filename}_Vs'.replace(comp, 'pyrolite'))
        Vphi_TH = lm.LayeredModel(f'{filename}_Vc'.replace(comp, 'pyrolite'))

        depths_above_200 = [i for i, l in enumerate(Vp_layer.layers) if l.depth < (2891 - 200)]
        for d in depths_above_200:
            Vp_layer.layers[d].cilm[:] = Vp_TH.layers[d].cilm[:]
            Vs_layer.layers[d].cilm[:] = Vs_TH.layers[d].cilm[:]
            Vphi_layer.layers[d].cilm[:] = Vphi_TH.layers[d].cilm[:]

    return Vp_layer, Vs_layer, Vphi_layer

    
def gridded_data(lon, lat, data):
    """
    Interpolates cartesian data onto a grid of equally spaced latitude and longitude
    points.

    Parameters
    -------
    lon: array_like (n)
        A list of longitude points at which the data points are located.
    lat: array_like (n)
        A list of latitude points at which the data points are located.
    data: array_like (n)
        The data stored at the points located on points given by the lon and lat
        array.

    Returns
    -------
    x: array_like (m,k)
        2-D array containing the x-coordinates of the interpolated grid.
    y: array_like (m,k)
        2-D array containing the y-coordinates of the interpolated grid.
    interpolated_data: array_like (m,k)
        2-D array containing the data points at the interpolated grid.
    """
    if max(lon) > 180:
        lon = ((lon+180)%360)-180
    interpolator = scipy.interpolate.NearestNDInterpolator((lon, lat), data, rescale=False)
    y, x = np.meshgrid(np.linspace(-90,90,181), np.linspace(-180,180,361))
    interpolated_data = interpolator(x, y)
    return x, y, interpolated_data


def rms(array, ax = None, type = None):
    """
    Calculates the root mean squared of an array in either cartesian or spherical
    space.

    Parameters
    -------
    array: array_like
        The data, either as an array or in spherical harmonics format.
    ax: int
        The axis along which to calculate the RMS.
    type: str
        The type of array data. "SH" if data is in spherical harmonics. If not "SH",
        the data is treated as an array.
        
    Returns
    -------
    : int or array_like
        The RMS values calculated.
    """    
    if type == "SH":
        return np.sqrt(np.sum((array ** 2) / (4 * np.pi), axis = ax))
    else:
        return np.sqrt(np.sum((array ** 2) / np.size(array), axis = ax))


def make_grids_with_def_filter(coefs, lat, lon, deg = None):
    """
    Expands spherical harmonic coefficients at given latitude and longitude points,
    with the option of expanding points for specific SH degrees.

    Parameters
    -------
    coefs: array_like (2,n,n)
        Spherical harmonic coefficients.
    lat: array_like (m)
        A list of latitude points to expand the data on.
    lon: array_like (m)
        A list of longitude points to expand the data on.
    deg: array_like
        The list of degrees to consider when expanding the coefficients. If deg is
        None, the expansion is done on all degrees.
        
    Returns
    -------
    : array_like (m)
        The expanded data points on the list of lat and lon values given.
    """
    new_coefs = np.zeros_like(coefs)

    if deg != None:
        for d in deg:
            new_coefs[:,d] = coefs[:,d]
    else:
        new_coefs = coefs

    return shtools.expand.MakeGridPoint(new_coefs, lat, lon, norm = 4)
```

```python
# Functions to plot graphs
def plot_two_phase_region(temperatures, phaseboundary, t_grid, min_model='SLB_2022', 
                          output_figure=output_figure, outdir=outdir):
    """
    Plots the two phase region of the bridgmanite-post-perovskite phase transition
    for a given mineralogical model.

    Parameters
    -------
    temperatures: array_like
        The list of temperatures at which the two phase region is evaluated at.
    phaseboundary: dict
        The phase boundaries evaluated for different compositions and ppv
        scenarios.
    t_grid: dict
        The temperature grid of the geodynamic model, which should include the
        thermal model and thermochemical model in keys 'TH' and 'TC' 
        respectively.
    min_model: str
        The mineralogical model used. Should either be "SLB_2022" or "SLB_2011".
    output_figure: bool
        If output_figure = True, the function will save the figure at the
        outdir directory.
    outdir: str
        The directory in which to output the figure.
    """
    CMB = burnman.seismic.PREM().pressure([2890*1000]) / 1e9
    
    fig, ax = plt.subplots(2, 2, figsize = (14,18))
    fig.subplots_adjust(wspace = 0, hspace = 0)
    depth_to_pressure = burnman.seismic.PREM().pressure(depth*1000)/1e9

    tH_mean = np.mean(t_grid['TH'], axis = 1)
    tH_err = np.percentile(t_grid['TH'], [10, 90], axis = 1)
    tC_mean = np.mean(t_grid['TC'], axis = 1)
    tC_err = np.percentile(t_grid['TC'], [10, 90], axis = 1)

    for axes, comp, c in zip(ax.reshape(-1), ['pyrolite','BMO','MORB','HC'], 
                                   ['b','r','orange','y']):
        if comp == 'pyrolite':
            t_mean = tH_mean
            t_err = tH_err
        else:
            t_mean = tC_mean
            t_err = tC_err
        axes.plot(t_mean, depth_to_pressure, 'k')
        axes.plot(t_err[0], depth_to_pressure, 'k--')
        axes.plot(t_err[1], depth_to_pressure, 'k--')
        axes.fill_between(temperatures, 
                        phaseboundary[f'{comp}_{min_model[-2:]}'].lowp/1e9, 
                        phaseboundary[f'{comp}_{min_model[-2:]}'].highp/1e9, 
                        color = c, alpha = 0.5, label = comp)
        axes.axhline(CMB, c = 'silver', linewidth = 5)
        axes.legend(fontsize = 15, framealpha = 0)
        axes.set_xlim([1000,4500])
        axes.set_ylim([150, 80])
        
        if min_model == 'SLB_2022':
            axes.text(1150, 125, 'bdg+pPv', fontsize = 15)
            axes.text(3500, 110, 'bdg', fontsize = 15)
            axes.text(3000, 142, 'pPv', fontsize = 15)
        else:
            axes.text(3500, 95, 'bdg', fontsize = 15)
            axes.text(1300, 130, 'pPv', fontsize = 15)
            if comp == 'MORB':
                axes.text(3800, 117, 'bdg+pPv', rotation = 35, fontsize = 15)
            elif comp == 'HC':
                axes.text(3800, 113, 'bdg+pPv', rotation = 35, fontsize = 15)
            elif comp == 'pyrolite':
                axes.text(3800, 133, 'bdg+pPv', rotation = -25, fontsize = 15)
            else:
                axes.text(3700, 135, 'bdg+pPv', rotation = -25, fontsize = 15)

    title_extension = ''
    if min_model == "SLB_2011":
        title_extension = ' using SLB2011'

    for axes in ax[0]:
        axes.set_xticklabels([], fontsize = 15)
    for axes in ax[-1]:
        axes.set_xlabel('Temperature (K)', fontsize = 15)
    ax[1,0].set_xticklabels(axes.get_xticklabels()[:-1], fontsize = 15)
    
    for rows in ax:
        rows[0].set_ylabel('Pressure (GPa)', fontsize = 15)
        rows[1].set_yticklabels([], fontsize = 15)
        ax2 = rows[1].twinx()
        ax2.set_ylim(rows[1].get_ylim())
        ax2.set_yticks(burnman.seismic.PREM().pressure(np.array([2000,2200,2400,2600,2800,2890])*1000) / 1e9, 
                    labels = ['2000', '2200', '2400','2600','2800', '2890'], fontsize = 15)
        ax2.set_ylabel('Depth (km)', fontsize = 15)
        
        for axes in rows:
            axes.tick_params(axis='both', labelsize = 15)
    ax[0,0].set_yticklabels(rows[0].get_yticklabels()[:-1], fontsize = 15)

    fig.suptitle(f'Bridgmanite-Post-perovskite phase boundary{title_extension}', fontsize = 23, y = 0.91)

    if output_figure:
        fig.savefig(os.path.join(outdir, f"two_phase_{min_model.replace('SLB_', '')}.png"), 
                    transparent = True, bbox_inches = 'tight')


def plot_SOLA_model(SOLA_model, titlesize = 35, labelsize = 22, ticklabelsize = 15):

    """
    Plots the resolving kernel, mean and uncertainties of $V_p$ and $V_s$ of the SOLA model.

    Parameters
    -------
    SOLA_model: gc.SOLAShell class object
        The SOLA model that will be plotted.
    titlesize: int
        The fontsize of the title.
    labelsize: int
        The fontsize of the axis labels.
    ticklabelsize: int
        The fontsize of the tick labels.
    """

    fig, ax = plt.subplots(3,3, figsize = (29, 17), width_ratios=[1.5, 6, 6], height_ratios=[6,6,1], subplot_kw = dict(projection = ccrs.Robinson(central_longitude=150)))


    tomoproxy.plotting.plot_shcoefs(SOLA_model.vp, quantity = f'dlnV$_p$ (%)', cmap = 'RdBu',
                                        show_cb=False,
                                        fig = fig,
                                        ax = ax[1,1],
                                        vmin = -0.5, vmax = 0.5, projection = ccrs.Robinson())
    tomoproxy.plotting.plot_shcoefs(SOLA_model.vs, quantity = f'dlnV$_s$ (%)', cmap = 'RdBu',
                                        show_cb=False,
                                        fig = fig,
                                        ax = ax[0,1],
                                        vmin = -1.0, vmax = 1.0, projection = ccrs.Robinson())

    tomoproxy.plotting.plot_err_coefs(SOLA_model.vp_err, quantity='$\sigma$ V$_p$',
                                show_cb=False,
                                fig = fig,
                                coast_color='k',
                                ax = ax[1,2],
                                vmax = 0.25, projection=ccrs.Robinson())
    tomoproxy.plotting.plot_err_coefs(SOLA_model.vs_err, quantity='$\sigma$ V$_s$',
                                show_cb=False,
                                coast_color='k',
                                fig = fig,
                                ax = ax[0,2],
                                vmax = 0.25, projection=ccrs.Robinson())

    ax[0,1].set_title('(b) Mean Model', fontsize = titlesize, y = 1.05)
    ax[0,2].set_title('(c) Uncertainties', fontsize = titlesize, y = 1.05)

    ax[1,1].set_xlabel('dlnV$_p$ (±0.5%)', fontsize = 1.3*labelsize, y = 0.98)
    ax[0,1].set_xlabel('dlnV$_s$ (±1.0%)', fontsize = 1.3*labelsize, y = 0.98)
    ax[1,1].set_xticks([])
    ax[0,1].set_xticks([])

    ax[1,2].set_xlabel('$\sigma$ V$_p$', fontsize = 1.3*labelsize, y = 0.98)
    ax[0,2].set_xlabel('$\sigma$ V$_s$', fontsize = 1.3*labelsize, y = 0.98)
    ax[1,2].set_xticks([])
    ax[0,2].set_xticks([])

    ax[-1,1] = plt.subplot(3,3,8, projection = 'rectilinear')
    seismic_cb = plt.cm.ScalarMappable(cmap = 'RdBu')
    seismic_cb.set_clim(-1,1)
    h = fig.colorbar(seismic_cb, cax = ax[-1,1], orientation = 'horizontal', ticks = [-1, 0, 1], spacing = 'proportional', extend = 'both')
    h.set_label('Velocity (%) with respect to radial average', size=labelsize)
    h.ax.set_xticklabels(['Lower', '0', 'Higher'], fontsize = ticklabelsize)

    ax[-1,2] = plt.subplot(3,3,9, projection = 'rectilinear')
    grey_cb = plt.cm.ScalarMappable(cmap = 'Greys')
    grey_cb.set_clim(0, 0.25)
    h2 = fig.colorbar(grey_cb, cax = ax[-1,2], orientation = 'horizontal', spacing = 'proportional', extend = 'max')
    h2.set_label('Velocity anomaly uncertainty (%)', size=labelsize)
    h2.ax.tick_params(labelsize = ticklabelsize)

    ax[0,0] = plt.subplot(3,3,(1,4), projection = 'rectilinear')
    ax[0,0].axvline(0, c = 'silver')
    ax[0,0].plot(SOLA_model.spline_vp, 6371 - SOLA_model.depths, 'b', linewidth = 2, label = 'd$\ln V_p$')
    ax[0,0].plot(SOLA_model.spline_vs, 6371 - SOLA_model.depths, 'r', linewidth = 2, label = 'd$\ln V_s$')
    ax[0,0].set_ylabel('Depth (km)', fontsize= labelsize)
    ax[0,0].set_ylim([2900, 0])
    ax[0,0].set_xlim([-0.02,0.07])
    ax[0,0].set_xticks([])
    ax[0,0].tick_params(axis = 'both', labelsize = ticklabelsize)
    ax[0,0].legend(fontsize = ticklabelsize)
    ax[0,0].set_title('(a) Resolution', fontsize = titlesize, y = 1.01)

    ax[-1,0].axis('off')

    return fig


def plot_filtered_synthetic_models(SOLA_model, synthetic_models, 
                                   v_type, cbar_lim, min_model, 
                                   central_longitude=0, 
                                   plot_type='presentation', 
                                   output_figure=output_figure,
                                   outdir=outdir):
    
    """
    Plots filtered synthetic velocity models, organised by post-perovskite 
    stability scenario and LLVP composition.
    
    Parameters
    -------
    SOLA_model: gc.SOLAShell class object
        The SOLA model that will be plotted.
    synthetic_models: dict
        The synthetic velocity models for different compositions and ppv
        scenarios.
    v_type: str
        The type of velocity being plotted. Must either be 'vp', 'vs' or 'vphi'.
    cbar_lim: int
        The maximum value (and minimum) to place the colorbar at. 
    min_model: str
        The mineralogical model used. Should either be "SLB_2022" or "SLB_2011".
    central_longitude: int
        The longitude that the plot will be centered around.
    plot_type: str
        Purpose of the plot. If plot_type = Presentation, the colorbars will
        be plotted vertically on the right. Otherwise, the colorbars are 
        plotted at the bottom.
    output_figure: bool
        If output_figure = True, the function will save the figure at the
        outdir directory.
    outdir: str
        The directory in which to output the figure.
    """

    assert v_type in ['vp', 'vs', 'vphi'], "Not valid velocity type"

    if plot_type == 'presentation':
        fig, ax = plt.subplots(4, 4, figsize = (23,13), subplot_kw = dict(projection = ccrs.Robinson(central_longitude = central_longitude)))
        fontsize = 25
        labelsize = 15
        ticklabelsize = 13
        fig.subplots_adjust(wspace = 0.08, hspace = 0.08)
        start_row = 0
        cb_orientation = 'vertical'
    else:
        fig, ax = plt.subplots(6, 3, figsize = (15,16), height_ratios=[6,6,6,6,6,1], subplot_kw = dict(projection = ccrs.Robinson(central_longitude = central_longitude)))
        fontsize = 20
        labelsize = 15
        ticklabelsize = 10
        start_row = 1
        cb_orientation = 'horizontal'

    # For plotting titles
    LLVP_comp = ["Pyrolite", "MORB", "BMO", "HC"]
    ppv_scenario = ["No pPv", "pPv", "Partial pPv"]

    # for dictionary keys
    comp_list = ['pyrolite', 'MORB', 'BMO', 'HC']
    ppv_list = ['noppv', 'ppv', 'partppv']

    # Declaring special axes for BG model and colorbars
    if plot_type == 'presentation':
        v_axis = plt.subplot(8,4, (8,12), projection = ccrs.Robinson(central_longitude = central_longitude))
        sigma_axis = plt.subplot(8,4, (16,20), projection = ccrs.Robinson(central_longitude = central_longitude))
        cb_v_axis = plt.subplot(8, 24, (140,188), projection = 'rectilinear')
        cb_sigma_axis = plt.subplot(8, 24, (143,191), projection = 'rectilinear')
    else:
        v_axis = plt.subplot(6,6, (2,3), projection = ccrs.Robinson(central_longitude = central_longitude))
        sigma_axis = plt.subplot(6,6, (4,5), projection = ccrs.Robinson(central_longitude = central_longitude))
        cb_v_axis = plt.subplot(6,3,(16,17), projection = 'rectilinear')
        cb_sigma_axis = plt.subplot(6,3,18, projection = 'rectilinear')

    tomoproxy.plotting.plot_shcoefs(getattr(SOLA_model, v_type), quantity = f'dlnV$_{v_type[1:]}$ (%)', cmap = 'RdBu',
                                        show_cb=False,
                                        fig = fig,
                                        ax = v_axis,
                                        vmin = -cbar_lim, vmax = cbar_lim, projection = ccrs.Robinson())
    v_axis.set_title('SOLA Mean Model', fontsize = fontsize)
    tomoproxy.plotting.plot_err_coefs(getattr(SOLA_model, v_type+'_err'), quantity = f'$\sigma$ V$_{v_type[1:]}$ (%)',
                                        show_cb=False,
                                        coast_color='k',
                                        fig = fig,
                                        ax = sigma_axis,
                                        vmax = 0.25, projection=ccrs.Robinson())
    sigma_axis.set_title('SOLA Model Uncertainties', fontsize = fontsize)

    for name, m in synthetic_models.items():
        split_name = name.split('_')
        comp = split_name[0]
        ppv = split_name[1]

        comp_index = comp_list.index(comp)
        ppv_index = ppv_list.index(ppv)

        if split_name[2] == min_model[-2:]:
            tomoproxy.plotting.plot_shcoefs(getattr(m, v_type), quantity = f'dV$_{v_type[1:]}$ (%)', cmap = 'RdBu',
                                            fig = fig,
                                            ax = ax[start_row + comp_index, ppv_index],
                                            vmin = -cbar_lim, vmax = cbar_lim, show_cb = False, projection = ccrs.Robinson())

    for i, axes in enumerate(ax[start_row, :len(ppv_scenario)]):
        axes.set_title(f'{ppv_scenario[i]}', fontsize = fontsize)
    for i, axes in enumerate(ax[start_row:start_row + len(LLVP_comp),0]):
        axes.set_ylabel(f'{LLVP_comp[i]}', fontsize = fontsize)
        axes.set_yticks([])

    trans = ax[0,0].get_transform()
    ax[0,0].plot([0,0],[10,10], color="k", transform=trans, clip_on=False)


    seismic_cb = plt.cm.ScalarMappable(cmap = 'RdBu')
    seismic_cb.set_clim(-cbar_lim, cbar_lim)
    h = fig.colorbar(seismic_cb, cax = cb_v_axis, orientation = cb_orientation, extend = 'both')
    h.set_label(f'dlnV$_{v_type[1:]}$ (%)', size=labelsize)
    h.ax.tick_params(labelsize = ticklabelsize)
    grey_cb = plt.cm.ScalarMappable(cmap = 'Greys')
    grey_cb.set_clim(0, 0.25)
    h2 = fig.colorbar(grey_cb, cax = cb_sigma_axis, orientation = cb_orientation, spacing = 'proportional', extend = 'max')
    h2.set_label(f'dlnV$_{v_type[1:]}$ uncertainty (%)', size=labelsize)
    h2.ax.tick_params(labelsize = ticklabelsize)
    if output_figure:
        fig.savefig(outdir + f'v{v_type[1:]}_{min_model[-2:]}.png', transparent = True)
        fig.savefig(outdir + f'v{v_type[1:]}_{min_model[-2:]}.eps', transparent = True)


def plot_RMS_velocities(synthetic_models,
                         vp_median, vs_median, 
                         vp_bounds_1, vs_bounds_1, 
                         vp_bounds_2, vs_bounds_2, 
                        ppv_frac=None, ppv_frac_xaxis=False,
                         min_model='SLB_2022', thermochemical_bottom=False,
                         output_figure=output_figure, outdir=outdir):
    
    """
    Plotting RMS dln velocities of the synthetic velocity models.

    Parameters
    -------
    synthetic_models: dict
        The synthetic velocity models for different compositions and ppv
        scenarios.
    vp_median: float
        The median Vp value of the SOLA model.
    vs_median: float
        The median Vs value of the SOLA model.
    vp_bounds_1: array_like (2)
        The lower and upper Vp values evaluated at 1 sigma uncertainty.
    vs_bounds_1: array_like (2)
        The lower and upper Vs values evaluated at 1 sigma uncertainty.
    vp_bounds_2: array_like (2)
        The lower and upper Vp values evaluated at 2 sigma uncertainty.
    vs_bounds_2: array_like (2)
        The lower and upper Vs values evaluated at 2 sigma uncertainty.
    ppv_frac: dict
        The depth-averaged pPv fraction of synthetic velocity models. Should
        have the same keys at the synthetic_models dictionary.
    ppv_frac_xaxis: bool
        If ppv_frac_xaxis=True, the x axis will be plotted as the pPv fraction
        of the synthetic velocity models given by ppv_frac. Otherwise, the
        x-axis spaces the models evenly.
    min_model: str
        The mineralogical model used. Should either be "SLB_2022" or "SLB_2011".
    thermochemical_bottm: bool
        If true, the diagram plots the synthetic models where only the bottom
        200km holds chemically distinct material. Note that the calculation
        is not done in this function, so the synthetic_model dictionary
        already needs to contain functions that only have chemically distinct
        material in the bottommost 200km.
    output_figure: bool
        If output_figure = True, the function will save the figure at the
        outdir directory.
    outdir: str
        The directory in which to output the figure.
    """

    assert min_model in ['SLB_2011', 'SLB_2022'], "Database not recognised. Must either be 2022 (SLB 2022) or 2011 (SLB 2011)"
    if ppv_frac_xaxis:
        assert ppv_frac is not None, "pPv fraction needed if ppv_frac_xaxis is True"

    suptitlesize = 16
    titlesize = 14
    labelsize = 11

    fig, ax = plt.subplots(1, 2, figsize = (9,6), sharex = True)
    fig.subplots_adjust(wspace = 0.25)
    comp_order_ppv = ['b', 'r', 'orange', 'y']
    marker= ['x','o', 'P']
    
    comp_list = ['pyrolite', 'BMO', 'MORB', 'HC']
    ppv_list = ['noppv', 'ppv', 'partppv']

    # Plot SOLA values + sigma uncertainties
    if ppv_frac_xaxis:
        lims = [-0.05, 0.85]
    else:
        lims = [-0.5, 12.5]
    ax[0].axhline(vp_median, c='k')
    ax[0].fill_between(lims, vp_bounds_2[0] * np.ones(2), vp_bounds_2[1] * np.ones(2), color = (0.9, 0.9, 0.9, 1))
    ax[0].fill_between(lims, vp_bounds_1[0] * np.ones(2), vp_bounds_1[1] * np.ones(2), color = (0.8, 0.8, 0.8, 1))
    ax[1].axhline(vs_median, c='k')
    ax[1].fill_between(lims, vs_bounds_2[0] * np.ones(2), vs_bounds_2[1] * np.ones(2), color = (0.9, 0.9, 0.9, 1))
    ax[1].fill_between(lims, vs_bounds_1[0] * np.ones(2), vs_bounds_1[1] * np.ones(2), color = (0.8, 0.8, 0.8, 1))

    # Only plotting newest database
    if min_model == 'SLB_2022':
        data_title = ''
    elif min_model == 'SLB_2011':
        data_title = ' using SLB2011'

    # for bottommost 200km thermochemical scenario
    if thermochemical_bottom:
        TC_title = ' of depth-restricted thermochemical LLVPs'
        TC_filename = '_200'
    else:
        TC_title = ''
        TC_filename = ''
        
    for name, m in synthetic_models.items():
        split_name = name.split('_')
        comp = split_name[0]
        ppv = split_name[1]

        comp_index = comp_list.index(comp)
        ppv_index = ppv_list.index(ppv)

        if split_name[2] == min_model[-2:]:    
            if ppv_frac_xaxis:
                ax[0].scatter(ppv_frac[name], rms(m.vp, type = 'SH'), c = comp_order_ppv[comp_index], marker = marker[ppv_index])
                ax[1].scatter(ppv_frac[name], rms(m.vs, type = 'SH'), c = comp_order_ppv[comp_index], marker = marker[ppv_index])
            else:
                x_loc = len(ppv_list) * comp_index + ppv_index 
                ax[0].scatter(x_loc, rms(m.vp, type = 'SH'), c = comp_order_ppv[comp_index], marker = marker[ppv_index])
                ax[1].scatter(x_loc, rms(m.vs, type = 'SH'), c = comp_order_ppv[comp_index], marker = marker[ppv_index])

    # Titles and labels
    ax[0].set_ylabel('dlnV$_p$ (%)', fontsize = labelsize)
    ax[1].set_ylabel('dlnV$_s$ (%)', fontsize = labelsize)
    ax[1].set_xlim(lims)
    ax[0].set_xticks([])
    if ppv_frac_xaxis:
        fig.text(0.5, 0.04, 'Measure of bdg-pPv co-occurrence', horizontalalignment='center', fontsize = labelsize)
    else:
        fig.text(0.5, 0.06, 'Synthetic Velocity Models', horizontalalignment='center', fontsize = labelsize)
    fig.suptitle(f'RMS velocities{TC_title}{data_title}', y = 0.97, fontsize = suptitlesize)
    ax[0].set_title('(a) dlnV$_p$', fontsize = titlesize)
    ax[1].set_title('(b) dlnV$_s$', fontsize = titlesize)


    # Legend
    ax[1].scatter([-1,-1],[0.5,0.5], label = '$\\bf{pPv}$ $\\bf{Model}$', c = 'none')
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'No pPv', c = 'k', marker = marker[0])
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'Partial pPv', c = 'k', marker = marker[2])
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'pPv', c = 'k', marker = marker[1])
    ax[1].scatter([-1,-1],[0.5,0.5], label = '\n$\\bf{LLVP}$ $\\bf{Comp}$.', c = 'none')
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'Pyrolite', c = 'b')
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'BMO', c = 'r')
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'MORB', c = 'orange')
    ax[1].scatter([-1,-1],[0.5,0.5], label = 'HC', c = 'y')

    # Right center legend
    fig.legend(loc = 'center right', bbox_to_anchor = (1.07, 0.5),fontsize = 10)

    if output_figure:
        fig.savefig(os.path.join(outdir, f'rms_velocity_{min_model[-2:]}{TC_filename}.png'), transparent = True, bbox_inches = 'tight')
        fig.savefig(os.path.join(outdir, f'rms_velocity_{min_model[-2:]}{TC_filename}.eps'), transparent = True, bbox_inches = 'tight')


def plot_ratios_and_corr(Rsp, rsc, 
                         R_median, corr_median, 
                         r_bounds_1, corr_bounds_1, 
                         r_bounds_2, corr_bounds_2, 
                         ppv_frac=None, ppv_frac_xaxis=False,
                         min_model='SLB_2022', 
                         output_type='poster',
                         output_figure=output_figure, outdir=outdir):
    
    """
    Plots the ratio $R_{s/p}$ and correlation $r_{s–c}$ of the synthetic 
    velocity models.

    Parameters
    -------
    Rsp: array_like
        The Rsp ratio for different compositions and ppv scenarios.
    rsc: array_like
        The rsc correlation for different compositions and ppv scenarios.
    R_median: float
        The median Rsp ratio of the SOLA model.
    corr_median: float
        The median rsc correlation of the SOLA model.
    r_bounds_1: array_like (2)
        The lower and upper Rsp ratios evaluated at 1 sigma uncertainty.
    corr_bounds_1: array_like (2)
        The lower and upper rsc correlations evaluated at 1 sigma 
        uncertainty.
    r_bounds_2: array_like (2)
        The lower and upper Rsp ratios evaluated at 2 sigma uncertainty.
    corr_bounds_2: array_like (2)
        The lower and upper rsc correlation evaluated at 2 sigma
        uncertainty.
    ppv_frac: dict
        The depth-averaged pPv fraction of synthetic velocity models. Should
        have the same keys at the synthetic_models dictionary.
    ppv_frac_xaxis: bool
        If ppv_frac_xaxis=True, the x axis will be plotted as the pPv fraction
        of the synthetic velocity models given by ppv_frac. Otherwise, the
        x-axis spaces the models evenly.
    min_model: str
        The mineralogical model used. Should either be "SLB_2022" or "SLB_2011".
    output_type: str
        Purpose of the plot. If output_type = presentation, the plot labels
        and symbols will be plotted at a larger size compared to output_type = 
        poster.
    output_figure: bool
        If output_figure = True, the function will save the figure at the
        outdir directory.
    outdir: str
        The directory in which to output the figure.
    """
    
    separate_columns = False
    assert min_model in ['SLB_2011', 'SLB_2022'], "Database not recognised. Must either be 2022 (SLB 2022) or 2011 (SLB 2011)"
    if ppv_frac_xaxis:
        assert ppv_frac is not None, "pPv fraction needed if ppv_frac_xaxis is True"

    if output_type == 'presentation':
        titlesize = 15
        labelsize = 15
        ticklabelsize = 12
        legendsize = 10
        size = 70
        fig, ax = plt.subplots(1,2, figsize = (9.5,6), sharex = True)
    else:
        suptitlesize = 15
        titlesize = 13
        labelsize = 12
        ticklabelsize = 10
        legendsize = 8
        size = 50
        fig, ax = plt.subplots(1,2, figsize = (10.1,6), sharex = True)
    fig.subplots_adjust(wspace = 0.35)

    edgesize = 1.7

    comp_order_ppv = ['b', 'b', 'b', 'r', 'r', 'r', 'orange', 'orange', 'orange', 'y', 'y', 'y']
    m_keys = ['pyrolite_noppv', 'pyrolite_ppv', 'pyrolite_partppv', 'BMO_noppv', 'BMO_ppv', 'BMO_partppv', 
            'MORB_noppv', 'MORB_ppv', 'MORB_partppv', 'HC_noppv', 'HC_ppv', 'HC_partppv']
    m= ['x','o','P']

    # Plotting Model
    if ppv_frac_xaxis:
        lims = [-0.15, 0.95]
    else:
        lims = [-0.5, 12.5]
    ax[0].axhline(R_median, c = 'k')
    ax[0].fill_between(lims, r_bounds_2[0] * np.ones(2), r_bounds_2[1] * np.ones(2), color = (0.9, 0.9, 0.9, 1))
    ax[0].fill_between(lims, r_bounds_1[0] * np.ones(2), r_bounds_1[1] * np.ones(2), color = (0.8, 0.8, 0.8, 1))
    ax[1].axhline(corr_median, c = 'k')
    ax[1].fill_between(lims, corr_bounds_2[0] * np.ones(2), corr_bounds_2[1] * np.ones(2), color = (0.9, 0.9, 0.9, 1))
    ax[1].fill_between(lims, corr_bounds_1[0] * np.ones(2), corr_bounds_1[1] * np.ones(2), color = (0.8, 0.8, 0.8, 1))

    if ppv_frac_xaxis:
        # make inset axis
        zm = ax[1].inset_axes([0,-0.7, 0.45, 1.2], transform=ax[1].transData)
        zm.set_xlim([0.78, 0.82])
        if min_model == 'SLB_2022':
            zm.set_ylim([-0.90, -0.78])
        elif min_model == 'SLB_2011':
            zm.set_ylim([-0.93, -0.78])

        zm.set_facecolor((0.97, 0.97, 0.97, 1))
        zm.set_xticks([])
        zm.set_yticks([])
        zm.fill_between(lims, corr_bounds_2[0] * np.ones(2), corr_bounds_2[1] * np.ones(2), color = (0.9, 0.9, 0.9, 1))

        ax[1].indicate_inset_zoom(zm, edgecolor = 'black')

    # Only plotting newest database
    if min_model == 'SLB_2022':
        start = 0
        data_title = ''
    elif min_model == 'SLB_2011':
        start = 12
        data_title = ' using SLB2011'

    if ppv_frac_xaxis:
        for i, (c, k) in enumerate(zip(comp_order_ppv, m_keys)):
            ax[0].scatter(ppv_frac[k+f'_{min_model[-2:]}'], Rsp[start+i,0], c = c, marker = m[i%3], linewidth = edgesize, s = size)
            ax[1].scatter(ppv_frac[k+f'_{min_model[-2:]}'], rsc[start+i,0], c = c, marker = m[i%3], linewidth = edgesize, s = size)
            zm.scatter(ppv_frac[k+f'_{min_model[-2:]}'], rsc[start+i,0], c = c, marker = m[i%3], linewidth = edgesize, s = size)
    else:
        for i, c in enumerate(comp_order_ppv):
            ax[0].scatter(i, Rsp[start+i,0], c = c, marker = m[i%3], linewidth = edgesize, s = size)
            ax[1].scatter(i, rsc[start+i,0], c = c, marker = m[i%3], linewidth = edgesize, s = size)


    ax[0].set_ylabel('$R_{s/p}$', fontsize = labelsize)
    ax[1].set_ylabel('$r_{s–c}$', fontsize = labelsize)

    #
    ax[0].set_title('(a) R$_{s/p}$', fontsize = titlesize)
    ax[1].set_title('(b) r$_{s–c}$', fontsize = titlesize)
    ax[1].set_xlim(lims)
    ax[0].set_xticks([])
    ax[0].set_ylim([1.4, 4.2])
    ax[1].set_ylim([-1.05, 1.05])
    ax[0].tick_params(axis = 'both', which = 'major', labelsize = ticklabelsize)
    ax[1].tick_params(axis = 'both', which = 'major', labelsize = ticklabelsize)

    # Legend
    ax[1].scatter([-1,-1],[0,0], label = '$\\bf{pPv}$ $\\bf{Model}$', c = 'none')
    ax[1].scatter([-1,-1],[0,0], label = 'No pPv', c = 'k', marker = m[0])
    ax[1].scatter([-1,-1],[0,0], label = 'Partial pPv', c = 'k', marker = m[2])
    ax[1].scatter([-1,-1],[0,0], label = 'pPv', c = 'k', marker = m[1])
    ax[1].scatter([-1,-1],[0,0], label = '\n$\\bf{LLVP}$ $\\bf{Comp}$.', c = 'none')
    ax[1].scatter([-1,-1],[0,0], label = 'Pyrolite', c = 'b')
    ax[1].scatter([-1,-1],[0,0], label = 'BMO', c = 'r')
    ax[1].scatter([-1,-1],[0,0], label = 'MORB', c = 'orange')
    ax[1].scatter([-1,-1],[0,0], label = 'HC', c = 'y')

    if ppv_frac_xaxis:
        fig.suptitle('$R_{s/p}$ and $r_{s–c}$ for filtered models'+f'{data_title}', y = 0.93, fontsize = suptitlesize)
    else:
        fig.suptitle('$R_{s/p}$ and $r_{s–c}$ for filtered models'+f'{data_title}', y = 0.97, fontsize = suptitlesize)

    if output_type == 'presentation':
        fig.legend(loc = 'center right', bbox_to_anchor = (1.08, 0.5), fontsize = legendsize)
        if ppv_frac_xaxis:
            figtext = 'Depth averaged pPv Fraction'
        else:
            figtext = 'Synthetic Velocity Models'
    else:
        if ppv_frac_xaxis:
            fig.legend(loc = 'lower center', bbox_to_anchor = (0.5, -0.04), ncols = 9, fontsize = legendsize)
            figtext = 'Measure of bdg-pPv co-occurrence'
        else:
            fig.legend(loc = 'center right', bbox_to_anchor = (1.03, 0.5), fontsize = legendsize)
            figtext = 'Synthetic Velocity Models'

    if separate_columns:
        ax[0].set_xlabel(figtext, fontsize = labelsize)
        ax[1].set_xlabel(figtext, fontsize = labelsize)
    else:
        if ppv_frac_xaxis:
            fig.text(0.5, 0.03, figtext, horizontalalignment='center', fontsize = labelsize)
        else:
            fig.text(0.5, 0.05, figtext, horizontalalignment='center', fontsize = labelsize)

    if output_figure:
        fig.savefig(os.path.join(outdir, f'ratios_filtered_{min_model[-2:]}.png'), transparent = True, bbox_inches = 'tight', dpi = 500)
        fig.savefig(os.path.join(outdir, f'ratios_filtered_{min_model[-2:]}.eps'), transparent = True, bbox_inches = 'tight', dpi = 500)


def plot_FS_R(FS_dist, min_model, output_figure=output_figure, outdir=outdir):
    """
    Plots the ratio $R_{s/p}$ for the fast and slow clusters of the synthetic 
    velocity models.

    Parameters
    -------
    FS_dist: dict
        The Rsp ratio for fast and slow clusters of the synthetic velocity
        models of different compositions and ppv scenarios.
    min_model: str
        The mineralogical model used. Should either be "SLB_2022" or "SLB_2011".
    output_figure: bool
        If output_figure = True, the function will save the figure at the
        outdir directory.
    outdir: str
        The directory in which to output the figure.
    """
    fig, ax = plt.subplots(4,3, figsize=(12,15))

    for i, comp in enumerate(['pyrolite', 'BMO', 'MORB', 'HC']):
        for j, ppv in enumerate(['noppv', 'ppv', 'partppv']):
            name = f'{comp}_{ppv}_{min_model[-2:]}'

            ax[i,j].hist(FS_dist[name][0], bins=np.arange(0,6,0.1), color = 'b', alpha = 0.65)
            ax[i,j].hist(FS_dist[name][1], bins=np.arange(0,6,0.1), color = 'r', alpha = 0.65)
            ax[i,j].axvline(np.mean(FS[0]), color = 'b', linestyle='--')
            ax[i,j].axvline(np.mean(FS[1]), color = 'r', linestyle='--')

    for axes,l in zip(ax[0],["No pPv", "pPv", "Part pPv"]):
        axes.set_title(l, fontsize = 15)

    for axes, l in zip(ax, ['Pyrolite', 'BMO', 'MORB', 'HC']):
        axes[0].set_ylabel(l, fontsize = 15)
    
    fig.supxlabel('R$_{s/p}$', y = 0.075, fontsize = 15)
    
    if min_model == 'SLB_2011':
        data_title = ' using SLB2011'
    else:
        data_title = ''

    fig.suptitle('Average R$_{s/p}$ for Fast and Slow Clusters'f'{data_title}', y = 0.94, fontsize = 20)

    if output_figure:
        fig.savefig(os.path.join(outdir, f"R_FS_geody_{min_model[-2:]}.eps"), transparent = True, bbox_inches = 'tight', dpi = 500)
        fig.savefig(os.path.join(outdir, f"R_FS_geody_{min_model[-2:]}.png"), transparent = True, bbox_inches = 'tight', dpi = 500)
```
